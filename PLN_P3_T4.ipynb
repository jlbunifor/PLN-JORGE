```python
import spacy
from spacy import displacy

# Carregar o modelo de português
nlp = spacy.load("pt_core_news_sm")

# Exemplo de frase
texto = "Estou estudando Processamento de Linguagem Natural com o professor Jorge Luiz"

# Processar o texto
doc = nlp(texto)

# Exibir no console as relações sintáticas (texto + dependência + cabeça)
for token in doc:
    print(f"{token.text:<12} {token.dep_:<15} {token.head.text:<12} ({token.pos_})")

# Visualizar a árvore sintática no navegador
displacy.serve(doc, style="dep")
```
