# Gerador de resposta

```python
# uma lista hst_mensagem é necessario
def gerador_de_resposta_historico(hst_mensagem):
   
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",        
        messages=hst_mensagem,
        n=1,temperature=0.5
    )
    return response.choices[0].message.content
```

# chatbot 

```python
def meu_chat_gpt_historico():
    # voce pode modificar o system template
    system_tempate="""
    você é um assistente conversacional que possui uma personalidade agradavel. Saiba que o usuario que conversa com voce é um aluno de Processamento de Linguagem Natural. Ajude-o! 
    
    """

    hst_mensagem=[{'role':'system','content':system_tempate}]
    while True: 
        user_input=input("USUARIO: ")
        if user_input.lower()!='sair':
            hst_mensagem.append({"role": "user", "content": user_input})
            respostaGPT=gerador_de_resposta_historico(hst_mensagem)
            hst_mensagem.append({"role": "assistant", "content": respostaGPT})
            print("GPT: ",respostaGPT)
            print("***********************")
        else:
            break
```
